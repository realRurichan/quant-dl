# 02 · Features
Technical indicators, sliding windows, chronological split.

```
import sys; sys.path.insert(0, '../src')
import numpy as np
from quant_dl.data import download
from quant_dl.features import FEATURE_COLUMNS, add_indicators, make_windows, chronological_split
```

```
df = add_indicators(download('AAPL', '2018-01-01', '2025-01-01'))
df[FEATURE_COLUMNS].describe()
```

```
WINDOW = 20
df = df.dropna()
X, y = make_windows(df, FEATURE_COLUMNS, WINDOW)
print(f'X: {X.shape}, y: {y.shape}')
```

```
(X_train, y_train), (X_val, y_val), (X_test, y_test) = chronological_split(X, y)
print(f'train={len(X_train)}, val={len(X_val)}, test={len(X_test)}')
print('first test date:', df.index[WINDOW + len(X_train) + len(X_val)])
```
